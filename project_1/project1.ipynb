```python
# Preserve the original data
df_raw = df.copy()

# Create a working DataFrame
df_clean = df.copy()

# Convert Date to datetime
df_clean["Date"] = pd.to_datetime(df_clean["Date"])

print("Original shape:", df_raw.shape)
print("Working shape:", df_clean.shape)
```

```python
missing_summary = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().sum() / len(df_clean) * 100
    ).round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_summary
```

```python
# Check the actual missing coupon records
df_clean.loc[
    df_clean["CouponCode"].isnull(),
    ["OrderID", "Product", "TotalPrice", "CouponCode"]
].head(10)
```

```python
df_clean["CouponCode"] = df_clean["CouponCode"].fillna(
    "No Coupon"
)

print("Remaining missing values:", df_clean.isnull().sum().sum())
print("CouponCode missing:", df_clean["CouponCode"].isnull().sum())
```

```python
print("Before treatment:", df_raw["CouponCode"].isnull().sum())
print("After treatment:", df_clean["CouponCode"].isnull().sum())

print("\nCoupon distribution:")
print(df_clean["CouponCode"].value_counts())
```

```python
def detect_outliers_iqr(data, column):
    q1 = data[column].quantile(0.25)
    q3 = data[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers)
    }
```

```python
numerical_columns = [
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]

outlier_results = pd.DataFrame({
    col: detect_outliers_iqr(df_clean, col)
    for col in numerical_columns
}).T

outlier_results
```

```python
result = detect_outliers_iqr(df_clean, "TotalPrice")

lower = result["Lower_Bound"]
upper = result["Upper_Bound"]

total_price_outliers = df_clean.loc[
    (df_clean["TotalPrice"] < lower) |
    (df_clean["TotalPrice"] > upper),
    ["OrderID", "Product", "Quantity", "UnitPrice", "TotalPrice"]
]

total_price_outliers
```

```python
total_price_stats = detect_outliers_iqr(
    df_clean, "TotalPrice"
)

df_clean["TotalPrice_Capped"] = df_clean["TotalPrice"].clip(
    lower=total_price_stats["Lower_Bound"],
    upper=total_price_stats["Upper_Bound"]
)
```

```python
print("Original maximum:",
      df_clean["TotalPrice"].max())

print("Capped maximum:",
      df_clean["TotalPrice_Capped"].max())

print("Rows retained:", len(df_clean))
```

```python
df_clean["OrderYear"] = df_clean["Date"].dt.year
```

```python
df_clean["OrderMonth"] = df_clean["Date"].dt.month
```

```python
df_clean["OrderQuarter"] = (
    "Q" + df_clean["Date"].dt.quarter.astype(str)
)
```

```python
df_clean["IsWeekend"] = (
    df_clean["Date"].dt.dayofweek >= 5
).astype(int)
```

```python
df_clean["CustomerOrderCount"] = (
    df_clean.groupby("CustomerID")["OrderID"]
    .transform("count")
)
```
